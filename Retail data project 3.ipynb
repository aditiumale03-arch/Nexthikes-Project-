## Outlier Detection 

### 1. Boxplots

```python
numeric_cols = EDA.select_dtypes(include='number').columns
colors = sns.color_palette("muted", len(numeric_cols))

for i, col in enumerate(numeric_cols):
    plt.figure(figsize=(9, 3))
    sns.boxplot(
        x=EDA[col],
        color=colors[i % len(colors)],
        linewidth=1.5,        # thinner border
        fliersize=4,          # smaller outlier markers
        boxprops=dict(alpha=0.85, edgecolor="gray")  # subtle border
    )
    plt.title(f"Boxplot of {col}", fontsize=13, fontweight='bold', color='navy')
    plt.xlabel(col, fontsize=11)
    plt.grid(axis='x', linestyle='--', alpha=0.4)
    plt.tight_layout()
    plt.show()

```

### 2. Interquartile Range (IQR) method 

```python
def iqr_outlier_summary(data, columns):
    results = []
    for col in columns:
        if col in data.columns and pd.api.types.is_numeric_dtype(data[col]):
            q1 = data[col].quantile(0.25)
            q3 = data[col].quantile(0.75)
            iqr = q3 - q1
            lower = q1 - 1.5 * iqr
            upper = q3 + 1.5 * iqr
            mask = (data[col] < lower) | (data[col] > upper)
            outlier_count = mask.sum()

            results.append({
                "Variable": col,
                "Q1": round(q1, 2),
                "Q3": round(q3, 2),
                "IQR": round(iqr, 2),
                "Lower Bound": round(lower, 2),
                "Upper Bound": round(upper, 2),
                "Outlier Count": outlier_count
            })
    return pd.DataFrame(results)

# Example usage
numeric_cols = EDA.select_dtypes(include='number').columns
iqr_summary = iqr_outlier_summary(EDA, numeric_cols)

# Display with professional gradient styling
print("IQR Outlier Summary:")
display(iqr_summary.style.background_gradient(cmap="Blues"))
```

## Skewness and Kurtosis Interpretation 

```python
# Skewness & Kurtosis — Interpretation

def interpret_skew(v):
    if v > 1:
        return "Highly positively skewed"
    if v > 0.5:
        return "Moderately positively skewed"
    if v < -1:
        return "Highly negatively skewed"
    if v < -0.5:
        return "Moderately negatively skewed"
    return "Approximately symmetric"

def interpret_kurt(v):
    return "High kurtosis / heavy tails" if v > 3 else "Normal-to-low kurtosis (thin tails)"

numeric_cols = EDA.select_dtypes(include='number').columns

for col in numeric_cols:
    s, k = EDA[col].skew(), EDA[col].kurt()
    print(f"{col:22s} skew={s:6.2f} -> {interpret_skew(s):32s} | kurt={k:6.2f} -> {interpret_kurt(k)}")
```
